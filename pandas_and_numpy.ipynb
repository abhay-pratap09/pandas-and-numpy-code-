````python
# Create DataFrame
data = {
    'Name': ['Abhay', 'Rahul', 'Aman', 'Priya', 'Abhay'],
    'Age': [20, 21, 19, 22, 20],
    'Marks': [85, 90, 75, 95, 85]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# 1. Basic Functions
print("\nFirst 2 Rows:")
print(df.head(2))

print("\nLast 2 Rows:")
print(df.tail(2))

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataFrame Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

print("\nTotal Elements:")
print(df.size)

# 2. Selecting Data
print("\nName Column:")
print(df['Name'])

print("\nFirst Row Using loc:")
print(df.loc[0])

print("\nFirst Row Using iloc:")
print(df.iloc[0])

print("\nStudents Older Than 20:")
print(df[df['Age'] > 20])

print("\nStudents With Marks Above 80:")
print(df.query('Marks > 80'))

# 3. Statistical Functions
print("\nTotal Marks:")
print(df['Marks'].sum())

print("\nAverage Marks:")
print(df['Marks'].mean())

print("\nMedian Marks:")
print(df['Marks'].median())

print("\nMode of Marks:")
print(df['Marks'].mode())

print("\nMinimum Marks:")
print(df['Marks'].min())

print("\nMaximum Marks:")
print(df['Marks'].max())

print("\nStandard Deviation:")
print(df['Marks'].std())

print("\nVariance:")
print(df['Marks'].var())

# 4. Sorting
print("\nSorted by Marks:")
print(df.sort_values(by='Marks'))

print("\nSorted by Age in Descending Order:")
print(df.sort_values(by='Age', ascending=False))

# 5. Counting Functions
print("\nCount of Non-Missing Values:")
print(df.count())

print("\nUnique Names:")
print(df['Name'].unique())

print("\nNumber of Unique Names:")
print(df['Name'].nunique())

print("\nFrequency of Each Name:")
print(df['Name'].value_counts())

# 6. Duplicate Functions
print("\nDuplicate Rows:")
print(df.duplicated())

print("\nDataFrame Without Duplicate Rows:")
print(df.drop_duplicates())

# 7. Adding and Removing Columns
df['Grade'] = ['B', 'A', 'C', 'A', 'B']

print("\nAfter Adding Grade Column:")
print(df)

df = df.drop(columns=['Grade'])

print("\nAfter Removing Grade Column:")
print(df)

# 8. Rename Column
df = df.rename(columns={'Marks': 'Score'})

print("\nAfter Renaming Marks to Score:")
print(df)

# 9. Apply Function
df['Score'] = df['Score'].apply(lambda x: x + 5)

print("\nAfter Adding 5 to Every Score:")
print(df)

# 10. GroupBy Function
print("\nAverage Score by Age:")
print(df.groupby('Age')['Score'].mean())

# 11. Missing Values
data2 = {
    'Name': ['Aman', 'Rahul', 'Priya'],
    'Marks': [80, None, 90]
}

df2 = pd.DataFrame(data2)

print("\nDataFrame With Missing Values:")
print(df2)

print("\nMissing Values Count:")
print(df2.isnull().sum())

print("\nAfter Filling Missing Values With 0:")
print(df2.fillna(0))

print("\nAfter Removing Rows With Missing Values:")
print(df2.dropna())

# 12. String Functions
print("\nNames in Uppercase:")
print(df['Name'].str.upper())

print("\nNames in Lowercase:")
print(df['Name'].str.lower())

print("\nLength of Names:")
print(df['Name'].str.len())

# 13. Combining DataFrames
df_a = pd.DataFrame({'Number': [1, 2]})
df_b = pd.DataFrame({'Number': [3, 4]})

print("\nConcatenated DataFrames:")
print(pd.concat([df_a, df_b], ignore_index=True))

print("\nProgram Executed Successfully!")
```
````
